```python
import json
from datetime import datetime
from pathlib import Path
import pandas as pd
```

```python

dossier_courant = Path.cwd()
chemin = 'default-cards.jsonl'
Dossier_json = dossier_courant / chemin
```

```python
data = []

# Lecture ligne par ligne
with open(Dossier_json, "r", encoding="utf-8") as f:
    for line in f:
        data.append(json.loads(line))

print(f"{len(data)} objets chargés.")

print(type(data))
print(type(data[0]))
```

On décide de garder que certains éléments

```python
mainDictList=[]

for json in data:
    tempDict={}
    tempDict["id"]=json["id"]

    if "oracle_id" in json.keys():
        tempDict["oracle_id"]=json["oracle_id"]
    else : 
        tempDict["oracle_id"]= None

    tempDict["name"]=json["name"]
    tempDict["lang"]=json["lang"]
    tempDict["released_at"]=json["released_at"]
    tempDict["layout"]=json["layout"]

    if "mana_cost" in json.keys():
        tempDict["mana_cost"]=json["mana_cost"]
    else : 
        tempDict["mana_cost"]= None

    if "cmc" in json.keys():
        tempDict["cmc"]=json["cmc"]
    else : 
        tempDict["cmc"]= None

    if "type_line" in json.keys():
        tempDict["type_line"]=json["type_line"]
    else : 
        tempDict["type_line"] = None

    if "oracle_text" in json.keys():
        tempDict["oracle_text"]=json["oracle_text"]
    else : 
        tempDict["oracle_text"]=None
    if "colors" in json.keys():
        tempDict["colors"]=json["colors"]
    else : 
        tempDict["colors"]=None
    tempDict["color_identity"]=json["color_identity"]
    tempDict["keywords"]=json["keywords"]
    if "produced_mana" in json.keys():
        tempDict["produced_mana"]=json["produced_mana"]
    else :
        tempDict["produced_mana"]=None
    tempDict["legalities"]=json["legalities"]
    tempDict["games"]=json["games"]
    tempDict["reserved"]=json["reserved"]
    tempDict["game_changer"]=json["game_changer"]
    tempDict["foil"]=json["foil"]
    tempDict["nonfoil"]=json["nonfoil"]
    tempDict["finishes"]=json["finishes"]
    tempDict["promo"]=json["promo"]
    tempDict["reprint"]=json["reprint"]
    tempDict["variation"]=json["variation"]
    if "card_faces" in json.keys():
        tempDict["card_faces"]=json["card_faces"]
    else :
        tempDict["card_faces"]= None
    if "defense" in json.keys():
        tempDict["defense"]=json["defense"]
    else : 
        tempDict["defense"]=None
    if "edhrec_rank" in json.keys():
        tempDict["edhrec_rank"]=json["edhrec_rank"]
    else : 
        tempDict["edhrec_rank"]=None

    if "loyalty" in json.keys():
        tempDict["loyalty"]=json["loyalty"]
    else : 
        tempDict["loyalty"]=None
    
    if "power" in json.keys():
        tempDict["power"]=json["power"]
    else : 
        tempDict["power"]=None
    if "toughness" in json.keys():
        tempDict["toughness"]=json["toughness"]
    else : 
        tempDict["toughness"]=None
    tempDict["set"]=json["set"]
    tempDict["set_name"]=json["set_name"]
    tempDict["set_type"]=json["set_type"]
    tempDict["collector_number"]=json["collector_number"]
    tempDict["digital"]=json["digital"]
    tempDict["rarity"]=json["rarity"]
    tempDict["artist"]=json["artist"]
    tempDict["border_color"]=json["border_color"]
    tempDict["full_art"]=json["full_art"]
    tempDict["textless"]=json["textless"]
    tempDict["booster"]=json["booster"]
    tempDict["story_spotlight"]=json["story_spotlight"]
    tempDict["prices"]=json["prices"]

    mainDictList.append(tempDict)

print(len(mainDictList))
print(len(mainDictList[0]))
```

Transformer la liste de dictionnaire en liste de liste

```python
df_fullset = pd.DataFrame(mainDictList)

print(df_fullset.head())
```

On analyse les données manquantes

```python
def nettoyer_valeur(val):
    if val == "" or (isinstance(val, (list, tuple)) and len(val) == 0):
        return None
    return val

df_fullset = df_fullset.map(nettoyer_valeur)

nb_none = df_fullset.isna().sum().sum()
print(nb_none)

nb_none_par_colonne  = df_fullset.isna().sum()
print(nb_none_par_colonne)

```

On va maintenant retirer toutes les lignes qui ne nous interessent pas : 
- ceux qui ne sont pas en "paper" dans games
- les planes (s'il y en a)
- les cartes paraissant après le 1er octobre
- pas légal en commander

```python
df_temp = df_fullset.copy()
```

```python
# Enlever cartes paraissant après le 1er octobre 2026
print("Taille du data-set au début des modifications : ", df_fullset.shape)
long_original = df_fullset.shape[0]

df_temp["released_at"] = pd.to_datetime(df_temp["released_at"])
df_temp = df_temp[df_temp["released_at"] <= "2026-10-01"]

print("Taille du data-set moins cartes publiée après le 1er octobre : ", df_temp.shape)
long_date = df_temp.shape[0]
print("Nombre de cartes retirées : ", long_original - long_date)
```

```python
games_diff = []
for k in df_fullset["games"]:
    if k not in games_diff :
        games_diff.append(k)

print(games_diff)
```

```python
# Enlever les cartes qui n'ont pas paper dans le nom donc aussi celle avec None

df_temp = df_temp[df_temp["games"].apply(lambda x: "paper" in x if isinstance(x, list) else False)]

print("Taille du data-set moins cartes publiée après le 1er octobre : ", df_temp.shape)
long_games = df_temp.shape[0]
print("Nombre de cartes retirées : ", long_date - long_games)
```

```python
# Enlever les cartes planar et tokens et schemes

nb_token = df_temp["type_line"].apply(lambda x: "token" in x.lower() if isinstance(x, str) else False).sum()
print(nb_token)

nb_planar = df_temp["type_line"].apply(lambda x: "plane" in x.lower() if isinstance(x, str) else False).sum()
print(nb_planar)

nb_schemes = df_temp["type_line"].apply(lambda x: "scheme" in x.lower() if isinstance(x, str) else False).sum()
print(nb_schemes)

# On garde uniquement les lignes qui NE CONTIENNENT PAS "token"
df_temp = df_temp[~df_temp["type_line"].str.contains("token", case=False, na=False)]
df_temp = df_temp[~df_temp["type_line"].str.contains("plane", case=False, na=False)]
df_temp = df_temp[~df_temp["type_line"].str.contains("scheme", case=False, na=False)]

print("Taille du data-set moins cartes publiée après le 1er octobre : ", df_temp.shape)
long_type = df_temp.shape[0]
print("Nombre de cartes retirées : ", long_games - long_type)
```
